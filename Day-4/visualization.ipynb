```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

%matplotlib inline
```

```python
df = pd.read_csv('/content/dataset.csv')

df.head()
```

```python
df = df.drop_duplicates()
```

```python
numerical_columns = df.select_dtypes(include=np.number)

numerical_columns.columns
```

```python
categorical_columns = df.select_dtypes(include='object')

categorical_columns.columns
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    numerical_columns.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(10,6))

df['track_genre'].value_counts().head(10).plot(
    kind='bar',
    color='skyblue'
)

plt.title("Top 10 Track Genres")
plt.xlabel("Genre")
plt.ylabel("Count")
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("fig1.png")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.histplot(
    df['popularity'],
    bins=20,
    kde=True
)

plt.title("Popularity Distribution")

plt.savefig("fig2.png")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x='danceability',
    y='energy',
    alpha=0.5
)

plt.title("Danceability vs Energy")

plt.savefig("fig3.png")

plt.show()
```

```python
top = df.groupby("track_genre")["popularity"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top.values,
    y=top.index
)

plt.title("Average Popularity by Genre")

plt.savefig("fig4.png")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(
    x=df['tempo']
)

plt.title("Tempo Distribution")

plt.savefig("fig5.png")

plt.show()
```

```python
plt.figure(figsize=(6,6))

df['explicit'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.ylabel("")

plt.title("Explicit vs Non-Explicit Songs")

plt.savefig("fig6.png")

plt.show()
```
